```python
# ==========================================
# EXPERIMENT 1: TEMPERATURE
# ==========================================

import time

prompt = "Write one short sentence about a student learning technology."

temperatures = [0.0, 0.7, 1.5]

for temperature in temperatures:

    print("\n" + "=" * 60)
    print("TEMPERATURE =", temperature)
    print("=" * 60)

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt,
            config=types.GenerateContentConfig(
                temperature=temperature,
                max_output_tokens=50
            )
        )

        print("Response:", response.text)

    except Exception as e:
        print("ERROR:", e)

    time.sleep(5)
```

```python
model_info = client.models.get(
    model="gemini-3.5-flash-lite"
)

print("Model:", model_info.name)
print("Temperature:", model_info.temperature)
print("Top P:", model_info.top_p)
print("Top K:", model_info.top_k)
```

```python
# ==========================================
# EXPERIMENT 2: topP
# ==========================================

import time

prompt = "Write one short sentence about a student learning technology."

top_p_values = [0.1, 1.0]

for top_p in top_p_values:

    print("\n" + "=" * 60)
    print("topP =", top_p)
    print("=" * 60)

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt,
            config=types.GenerateContentConfig(
                top_p=top_p,
                max_output_tokens=50,
                thinking_config=types.ThinkingConfig(
                    thinking_level="minimal"
                )
            )
        )

        print("Response:", response.text)

    except Exception as e:
        print("ERROR:", e)

    time.sleep(5)
```

```python
model_info = client.models.get(
    model="gemini-3.5-flash-lite"
)

print("Model:", model_info.name)
print("TopK:", model_info.top_k)
print("TopP:", model_info.top_p)
print("Temperature:", model_info.temperature)
```

```python
# ==========================================
# EXPERIMENT 3: topK
# ==========================================

import time

prompt = "Complete this sentence: A student learning technology can"

top_k_values = [1, 40]

for top_k in top_k_values:

    print("\n" + "=" * 60)
    print("topK =", top_k)
    print("=" * 60)

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt,
            config=types.GenerateContentConfig(
                top_k=top_k,
                max_output_tokens=50
            )
        )

        print("Response:", response.text)

    except Exception as e:
        print("ERROR:", e)

    time.sleep(5)
```

Observation: topK controls the maximum number of high-probability tokens considered during sampling. With topK = 1, the candidate pool is extremely restricted. With topK = 40, up to 40 of the most probable tokens can be considered. Unlike topP, which uses a cumulative probability threshold, topK uses a fixed number of candidate tokens.

```python
# ==========================================
# EXPERIMENT 4: maxOutputTokens
# ==========================================

prompt = """
Explain the development of computers from early computers
to modern artificial intelligence. Include important stages
and explain each stage briefly.
"""

token_limits = [20, 50, 100]

for limit in token_limits:

    print("\n" + "=" * 60)
    print("maxOutputTokens =", limit)
    print("=" * 60)

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt,
            config=types.GenerateContentConfig(
                temperature=0.7,
                max_output_tokens=limit
            )
        )

        print(response.text)

    except Exception as e:
        print("ERROR:", e)
```

Observation: maxOutputTokens controls the maximum number of output tokens that the model can generate. When the value is small, the response has less space and may stop earlier. Increasing the value gives the model more room to produce a longer response. In this experiment, 20, 50, and 100 tokens were tested using the same prompt.

Experiment 5 — stopSequences

```python
# ==========================================
# EXPERIMENT 5: STOP SEQUENCES
# ==========================================

prompt = """Give me three programming languages.
After the third language, write the word END.
"""

print("WITHOUT STOP SEQUENCE")
print("=" * 50)

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config=types.GenerateContentConfig(
        max_output_tokens=100
    )
)

print(response.text)


print("\nWITH STOP SEQUENCE: END")
print("=" * 50)

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config=types.GenerateContentConfig(
        max_output_tokens=100,
        stop_sequences=["END"]
    )
)

print(response.text)
```

Observation: The stop sequence provides a predefined stopping point for text generation. In this experiment, END was used as the stop sequence. Without a stop sequence, the model can continue generating according to the prompt. With END configured as a stop sequence, generation stops when that sequence is reached, and the stop sequence is not included in the returned response.

```python
# ==========================================
# EXPERIMENT 7: STREAMING
# ==========================================

prompt = """
Explain artificial intelligence in simple words
for a beginner.
"""

print("STREAMING RESPONSE")
print("=" * 60)

stream = client.models.generate_content_stream(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config=types.GenerateContentConfig(
        max_output_tokens=100
    )
)

for chunk in stream:
    print(chunk.text, end="", flush=True)

print("\n")
```

```python
# ==========================================
# EXPERIMENT 7: STREAMING
# ==========================================

prompt = """
Explain artificial intelligence in simple words
for a beginner.
"""

print("STREAMING RESPONSE")
print("=" * 60)

stream = client.models.generate_content_stream(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config=types.GenerateContentConfig(
        max_output_tokens=100
    )
)

for chunk in stream:
    if chunk.text:
        print(chunk.text, end="", flush=True)

print("\n")
```

Observation: Streaming returns the generated response incrementally in chunks instead of waiting for the complete response. This allows text to be displayed while the model is still generating, making applications more responsive.
